```python
import pandas as pd

# List of student names to cycle through for 50 rows
sample_names = ["Rahul", "Priya", "Amit", "Neha", "Karan", "Riya", "Mohit", "Anjali", "Vivek", "Pooja"]

# Generate 50 rows of data (RollNo, Name, and 98 subject marks)
data = []
for i in range(50):
    roll_no = 101 + i
    name = sample_names[i % len(sample_names)]
    # Generate realistic marks (60 to 100) using basic arithmetic
    marks = [(60 + (i * 7 + j * 3) % 41) for j in range(98)]
    data.append([roll_no, name] + marks)

# Define 100 column names (RollNo + Name + Sub_1 to Sub_98)
columns = ["RollNo", "Name"] + [f"Sub_{j+1}" for j in range(98)]

# Create DataFrame
df = pd.DataFrame(data, columns=columns)

# Display head, tail, and shape
print("--- HEAD ---")
print(df.head())

print("\n--- TAIL ---")
print(df.tail())

print("\n--- SHAPE ---")
print(df.shape)
```

```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

num_records = 1000

# Sample data pools
first_names = ["Aarav", "Ananya", "Rohan", "Priya", "Kabir", "Neha", "Rahul", "Isha", "Karan", "Sanya", 
               "Vikram", "Riya", "Amit", "Pooja", "Vivek", "Meera", "Arjun", "Kavya", "Krishna", "Sai"]
last_names = ["Sharma", "Verma", "Gupta", "Mehta", "Singh", "Kumar", "Patel", "Jain", "Agarwal", "Yadav"]
departments = ["HR", "IT", "Finance", "Marketing", "Sales", "Support", "Operations"]

# Generate base data
emp_ids = [f"EMP{i:04d}" for i in range(1, num_records + 1)]
names = [f"{np.random.choice(first_names)} {np.random.choice(last_names)}" for _ in range(num_records)]
depts = [np.random.choice(departments) for _ in range(num_records)]
ages = np.random.randint(22, 61, size=num_records).astype(float)
salaries = np.random.randint(30000, 150001, size=num_records).astype(float)

# Create DataFrame
df = pd.DataFrame({
    "Employee ID": emp_ids,
    "Name": names,
    "Department": depts,
    "Age": ages,
    "Salary": salaries
})

# Inject ~5% missing values (np.nan) randomly into Name, Department, Age, and Salary
for col in ["Name", "Department", "Age", "Salary"]:
    mask = np.random.rand(num_records) < 0.05
    df.loc[mask, col] = np.nan

# Save to CSV
df.to_csv("company_missing_data.csv", index=False)

# Display sample output and missing value summary
print("--- FIRST 10 ROWS ---")
print(df.head(10))

print("\n--- MISSING VALUE COUNT PER COLUMN ---")
print(df.isna().sum())
```
