```python
rows = 5

for i in range(1, rows + 1):
    print(" " * (rows - i) + "*" * (2 * i - 1))
```

```python
rows = 5
pyramid = []

for i in range(rows, 0, -1):
    row = " " * (rows - i) + "*" * (2 * i - 1)
    pyramid.append(row)

for row in pyramid:
    print(row)
```

```python
score = float(input("Enter your score (0 to 100): "))

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print("Your grade is:", grade)
```
