```python
# Parent Class (Base Class)

class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_person(self):
        print("Name :", self.name)
        print("Age  :", self.age)


# Child Class (Derived Class)

class Student(Person):

    def __init__(self, name, age, roll_no, marks):
        # Call Parent Constructor
        super().__init__(name, age)

        # Child Class Variables
        self.roll_no = roll_no
        self.marks = marks

    def show_student(self):
        print("Roll Number :", self.roll_no)
        print("Marks       :", self.marks)


# Create Object

student1 = Student("Ayukt Yadav", 19, 26, 92)


# Display Data

print("Student Information")
print("--------------------")

student1.show_person()      # Parent Class Method
student1.show_student()     # Child Class Method
```

```python
marks = [85, 92, 78, 90, 88]

first = second = float('-inf')

for mark in marks:
    if mark > first:
        second = first
        first = mark
    elif mark > second and mark != first:
        second = mark

print("Second Highest Marks:", second)
```
